## 🧹**Data Cleaning** <a class="anchor" id="data_cleaning_section"></a>

```python
concert_details = concert_details.drop(columns=["Ref."],axis=1)
concert_details = concert_details.drop(columns=["All Time Peak"],axis=1)
concert_details
```

```python
concert_details = concert_details.drop(columns=["Peak"],axis=1)
concert_details
```

```python
concert_details = concert_details.rename(columns={"Adjusted gross (in 2022 dollars)":"Adjusted gross","Year(s)":"Year","Shows":"Number of shows"})
concert_details
```

```python
concert_details.columns
```

### **Renaming column names**

```python
concert_details = concert_details.rename(columns={concert_details.columns[2]:"Adjusted gross in USD",concert_details.columns[1]:"Actual gross in USD",concert_details.columns[7]:"Average gross in USD"})
concert_details
```

### **Changing position of columns in dataset**

```python
popped_artist_data = concert_details.pop("Artist")
concert_details.insert(1,"Artist",popped_artist_data)
popped_tour_title_data = concert_details.pop("Tour title")
concert_details.insert(2,"Tour title",popped_tour_title_data)
popped_average_gross_in_usd_data = concert_details.pop("Average gross in USD")
concert_details.insert(5,"Average gross in USD",popped_average_gross_in_usd_data)
concert_details.head()
```

```python
concert_details.info()
```

```python
concert_details["Tour title"] = concert_details["Tour title"].apply(lambda tourname:tourname[0:tourname.find("[")] if tourname.find("[")>=0 else tourname)
concert_details
```

```python
concert_details["Tour title"] = concert_details["Tour title"].str.replace("‡","")
concert_details["Tour title"] = concert_details["Tour title"].str.replace("†","")
concert_details["Tour title"] = concert_details["Tour title"].str.replace("*","")
concert_details["Tour title"] = concert_details["Tour title"].str.strip(" ")
concert_details
```

```python
concert_details["Actual gross in USD"] = concert_details["Actual gross in USD"].apply(lambda actual_gross:actual_gross[0:actual_gross.find("[")] if actual_gross.find("[")>=0 else actual_gross)
concert_details
```

```python
concert_details["Actual gross in USD"] = concert_details["Actual gross in USD"].str.replace("$","")
concert_details["Actual gross in USD"] = concert_details["Actual gross in USD"].str.replace(",","")
concert_details["Adjusted gross in USD"] = concert_details["Adjusted gross in USD"].str.replace("$","")
concert_details["Adjusted gross in USD"] = concert_details["Adjusted gross in USD"].str.replace(",","")
concert_details["Average gross in USD"] = concert_details["Average gross in USD"].str.replace("$","")
concert_details["Average gross in USD"] = concert_details["Average gross in USD"].str.replace(",","")
concert_details
```

### **Datatype checking**

```python
concert_details.dtypes
```

### **Changing datatype of columns**

```python
concert_details["Actual gross in USD"] = concert_details["Actual gross in USD"].astype("int")
concert_details["Adjusted gross in USD"] = concert_details["Adjusted gross in USD"].astype("int")
concert_details["Average gross in USD"] = concert_details["Average gross in USD"].astype("int")
concert_details.head()
```

```python
concert_details.info()
```

### **Checking number of unique**

### **Creating new column by splitting data**

```python
concert_details["start_year"] = concert_details["Year"].apply(lambda year:year[0:4])
concert_details.head()
```

```python
concert_details["end_year"] = concert_details["Year"].apply(lambda year:year[-4:])
concert_details.head()
```

```python
concert_details.drop(columns=["Year"],axis=1,inplace=True)
concert_details.head()
```

### **Standardizing column names**

```python
concert_details.columns = concert_details.columns.str.lower()
concert_details.head()
```

```python
concert_details.columns = concert_details.columns.str.replace(" ","_")
concert_details.columns = concert_details.columns.str.strip()
concert_details.head()
```

#### **Observation:**
- Column names are standardized to lower case with underscore for consistency and ease of analytics.

```python
concert_details.dtypes
```

####**Change datatype of years to integer**

```python
concert_details["start_year"] = concert_details["start_year"].astype("int")
concert_details["end_year"] = concert_details["end_year"].astype("int")
concert_details.dtypes
```

```python
concert_details.head()
```

```python
concert_details["duration_in_years"] = concert_details["end_year"]-concert_details["start_year"]
concert_details.head()
```

## 📊 **Data Visualization** <a class="anchor" id="data_visualization_section"></a>

### **Univariate analysis** <a class="anchor" id="univariate_analysis_section"></a>

```python
taylor_swift_concerts = concert_details[concert_details["artist"]=="Taylor Swift"]
plt.boxplot(taylor_swift_concerts["actual_gross_in_usd"],tick_labels=["Taylor Swift"]);
plt.title("Actual gross of tours performed by Taylor Swift")
plt.ylabel("Actual Gross (Millions USD)");
plt.savefig("/kaggle/working/plots/Actual gross of tours performed by Taylor Swift.png")
```

#### 💡**Insights:**
- The median is near to the first quartile, which means that most of the tours by Taylor Swift has less actual gross amount compared to the higher actual gross amount by Taylor Swift.
- The distribution appears right-skewed because of the longer upper whisker.

```python
top_5_highest_actual_grossing = concert_details.sort_values(by="actual_gross_in_usd",ascending=False).head(5)
plt.pie(top_5_highest_actual_grossing["actual_gross_in_usd"],labels=top_5_highest_actual_grossing["tour_title"]);
plt.title("Top 5 tours that made highest actual gross in USD");
plt.savefig("/kaggle/working/plots/Top_5_tours_that_made_highest_actual_gross_in_USD.png")
```

#### 💡**Insights:**
- The Eras Tour has made the highest actual gross among the top 5 highest actual grossing concert tours.
- The Renaissance World Tour has got the second highest actual gross among the top 5 highest actual grossing concert tours.
- The Reputation Stadium Tour has made the least actual gross among all the top 5 highest actual gross tours.

```python
madonna_concerts = concert_details[concert_details["artist"]=="Madonna"]
beyonce_concerts = concert_details[concert_details["artist"]=="Beyoncé"]
pink_concerts = concert_details[concert_details["artist"]=="Pink"]
plt.boxplot([taylor_swift_concerts["actual_gross_in_usd"],madonna_concerts["actual_gross_in_usd"],beyonce_concerts["actual_gross_in_usd"],pink_concerts["actual_gross_in_usd"]],tick_labels=["Taylor Swift","Madonna","Beyoncé","Pink"]);
plt.title("Artists vs Actual gross in USD");
plt.ylabel("Actual Gross (Millions USD)");
plt.savefig("/kaggle/working/plots/Artists vs Actual gross in USD.png")
```

#### 💡**Insights:**
- The spread of actual gross earned by the tours of Taylor Swift is more than the actual gross earned by the tours of Beyonce, Madonna, and Pink.

```python
plt.boxplot(concert_details["number_of_shows"])
plt.title("Number of shows performed by artists")
plt.ylabel("Number of shows");
plt.savefig("/kaggle/working/plots/number_of_shows_performed_by_artists.png")
```

#### 💡**Insights:**
- The median value is near to the lower quartile which means that most number of shows are below 100.
- The presence of outlier above the upper whisker indicates that one of the artists exceeded the other artists in the number of shows performed which is more than 300.

### **Bivariate analysis** <a class="anchor" id="bivariate_analysis_section"></a>

```python
plt.figure(figsize=(11,5))
plt.bar(concert_details["artist"].value_counts().index,concert_details["artist"].value_counts().values,width=0.3,color="purple")
plt.title("Total number of tours done by artist")
plt.xlabel("Artist name")
plt.ylabel("Number of Tours")
plt.savefig("/kaggle/working/plots/Total number of tours done by artist.png")
```

#### 💡**Insights:**
- Taylor Swift and Madonna have performed 4 tours which is high than other artists.
- Celine Dion, Katy Perry, Cher, and Adele have performed less number of tours.

```python
plt.figure(figsize=(12,7))
plt.plot(concert_details["rank"],concert_details["actual_gross_in_usd"],marker="o",color="red",label="Actual gross")
plt.plot(concert_details["rank"],concert_details["adjusted_gross_in_usd"],marker="o",color="green",label="Adjusted gross")
plt.plot(concert_details["rank"],concert_details["average_gross_in_usd"],marker="o",color="blue",label="Average gross")
plt.xticks(concert_details["rank"])
plt.xlabel("Rank of artist")
plt.ylabel("Gross amount (Millions USD)")
plt.legend();
plt.savefig("/kaggle/working/plots/Rank of artist.png")
```

#### 💡**Insights:**

- The actual gross is high for higher rank artists, and gets reduced when the rank of the artist is low.
- The adjusted gross is high for higher rank artists, and gets reduced when the rank of artist is low.
- The average gross of all ranks remains low.

```python
top_5_lowest_actual_grossing = concert_details.sort_values(by="actual_gross_in_usd").head(5)
plt.figure(figsize=(11,5))
plt.bar(top_5_lowest_actual_grossing["tour_title"],top_5_lowest_actual_grossing["actual_gross_in_usd"],width=0.3,color="red")
plt.title("Top 5 tours that made lowest actual gross in USD");
plt.xlabel("Tour titles")
plt.ylabel("Actual Gross (Millions USD)");
plt.savefig("/kaggle/working/plots/top_5_tours_that_made_lowest_actual_gross_in_usd.png")
```

#### 💡**Insights:**
- The lowest actual grossing tour is 'The Red Tour' which is the lowest among all the concert tours.

```python
plt.scatter(concert_details["number_of_shows"],concert_details["actual_gross_in_usd"],color="magenta",s=[50 for _ in range(20)]);
plt.title("Number of shows performed vs actual gross amount (Millions USD)")
plt.xlabel("Number of shows done by artists")
plt.ylabel("Actual gross earned (Millions USD)");
plt.savefig("/kaggle/working/plots/number_of_shows_performed_vs_actual_gross_amount_in_usd.png")
```

#### 💡**Insights:**
- Lesser number of shows performed by the artists earned more than the highest number of shows performed by the artists.

```python
total_number_of_shows_by_artists = concert_details.groupby("artist")["number_of_shows"].sum()
names_of_artists = np.array(total_number_of_shows_by_artists.index)
total_number_of_shows_by_each_artist = np.array([total_number_of_shows_by_artists[currname] for currname in names_of_artists])
plt.figure(figsize=(10,5))
plt.bar(names_of_artists,total_number_of_shows_by_each_artist,width=0.25,color="yellow",edgecolor="k")
plt.title("Artists vs Total number of shows performed")
plt.xlabel("Artist")
plt.ylabel("Total number of shows performed")
plt.savefig("/kaggle/working/plots/Artists_vs_Total_number_of_shows_performed.png")
```

#### 💡**Insights:**
- Cher, Lady Gaga, Madonna, and Pink has performed most of the total number of shows which is above 300.
- Adele and Celine Dion have performed less number of shows which is less than 150.

```python
plt.bar(concert_details["duration_in_years"],concert_details["average_gross_in_usd"],width=0.35,color="maroon")
plt.title("Duration of Tour vs Average gross earned in concerts (USD)")
plt.xlabel("Duration of tour in years")
plt.ylabel("Average gross (Millions USD)");
plt.savefig("/kaggle/working/plots/average_gross_in_usd_bar_chart.png")
```

#### 💡**Insights:**
- Higher the duration of concert in years, lower the average gross earned.
- The concerts with lesser duration like less than a year, 1 year have earned more average gross.

### **Multivariate analysis** <a class="anchor" id="multivariate_analysis_section"></a>

```python
corr_matrix = concert_details.corr(numeric_only=True)
seaborn.heatmap(corr_matrix,annot=True,cmap="Purples")
plt.title("Correlation Heatmap")
plt.tight_layout()
```

#### 💡**Insights**
- There is a dependency between the duration of the concert in years and the number of shows, because of strong positive correlation.
- There is no correlation between the rank of the artist and the gross amounts earned through the concerts, because there is weak negative correlation between them.
